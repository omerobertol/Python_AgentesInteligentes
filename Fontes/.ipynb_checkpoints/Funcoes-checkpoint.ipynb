```python
import pandas as pd

# criando o dataframe "ds"
ds = pd.read_csv("C:\Python_AgentesInteligentes\Fontes\Dados\census.csv")

idades = ds['age']

print(type(idades))
print()
print(idades)
```

```python
# mostrar somente os registros com valores menores que 18 anos
idades[idades < 18]
```

```python
def menorDe18(idade):
    if (idade < 18):
        idade = -1
    return(idade)
```

```python
novasIdades = idades.apply(menorDe18)
print(novasIdades[novasIdades < 18])
```

```python
novasIdades = idades.apply(lambda x: -1 if x <= 18 else x)
print(novasIdades[novasIdades <= 18])
```
