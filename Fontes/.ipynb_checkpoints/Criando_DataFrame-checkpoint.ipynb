```python
import pandas as pd
import numpy as np

print('Pandas:', pd.__version__)
print('NumPy:', np.__version__)
```

```python
# a partir de listas
dados = [['Joaquim', 4, 'Curitiba', 'PR'],
         ['Vanessa', 42, 'Pato Branco', 'PR'],
         ['Alzemiro', 92, 'Espumoso', 'RS'],
         ['Alice', 33, 'São Paulo', 'SP']]

colunas = ['Nome', 'Idade', 'Cidade', 'Estado']

# criando um DataFrame a partir de listas
df = pd.DataFrame(dados, columns=colunas)

print(type(df))
print()
print(df)
```

```python
# criando um DataFrame a partir de um dicionário de dados
dic = {'Nome' : ['Joaquim', 'Vanessa', 'Alzemiro', 'Alice'],	
       'Idade' : [4, 42, 92, 33],
       'Cidade' : ['Curitiba', 'Pato Branco', 'Espumoso', 'São Paulo'],
       'Estado' : ['PR', 'PR', 'RS', 'SP']}

print(type(dic))

df = pd.DataFrame(dic)
print(type(df))
print()
print(df)
```
