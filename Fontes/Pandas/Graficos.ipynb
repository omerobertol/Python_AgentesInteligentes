```python
import pandas as pd

df = pd.DataFrame({
    'vendedor': ['Ana', 'Joaquim', 'Carlos'],
    'vendas': [300.00, 450.00, 250.00]
})

print(type(df))
print()
print(df)
```

```python
df.plot(
    title = "Vendas x Vendedores",
    x = "vendedor",
    xlabel = "VENDEDOR",
    y = "vendas",
    ylabel = "VENDAS",
    kind = "bar",
    grid = True,
    legend=False
)
```
