```python
import pandas as pd
import numpy as np

nros = pd.Series([10, 11, np.nan, np.nan, 14, np.nan, 16])
print(nros)
```

```python
nros.isna()
```

```python
# qtde de registros com valores faltantes
nros.isna().sum()
```

```python
# apaga os registros com valores faltantes
nros.dropna(inplace=True)
```

```python
nros
```

```python
# substituir os valores faltantes pela media
nros.fillna(nros.mean(), inplace=True)
```
