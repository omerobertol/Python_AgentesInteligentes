```python
import pandas as pd

x = pd.Series([1, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'])

print(type(x))
print()
print(x) # 1a. Forma: mostrando toda a lista de uma só vez
```

```python
# 2a. Forma: percorrendo apenas os valores (item por item)
for item in x: # usando um for - each
    print(item)
print()    
```

```python
# 3a. Forma: percorrendo indices e valores
for i, valor in x.items():
    print(f"{i}: {valor}")
```

```python
# 4a. Forma: percorrendo usando índice padrão (numérico)
n = x.size
for i in range(n):
    print(x.iloc[i])
```

```python
# 5a. Forma: percorrendo usando o índice real
lista_indices = x.index.tolist()

print(lista_indices)
print()

for i in lista_indices:
    print(f"{i}: {x.loc[i]} {x[i]}")   
```
