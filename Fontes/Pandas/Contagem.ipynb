```python
import pandas as pd

# criando o dataframe "ds"
ds = pd.read_csv("C:\Python_AgentesInteligentes\Fontes\Dados\census.csv")

idades = ds['age']

idades.value_counts()
```

```python
paises_idades = pd.Series(data=ds['age'].values, index=ds['native-country'].values)

print(paises_idades)
print()
print(len(paises_idades), 'registros')
```

```python
# quantidade de registros por grupo
paises_contagem = paises_idades.groupby(level=0).count().sort_values()
print(paises_contagem)
```

```python
# idades médias por paises
idades_medias = paises_idades.groupby(level=0).mean()
idades_medias
```

```python
idades_medias = paises_idades.groupby(level=0).mean().sort_values()
idades_medias
```

```python
# mostra os 10 menores idades médias (agrupado por pais)
idades_medias = paises_idades.groupby(level=0).mean().sort_values().head(10)
idades_medias
```

```python
# mostra os 10 maiores idades médias (agrupado por pais)
idades_medias = paises_idades.groupby(level=0).mean().sort_values().tail(10)
idades_medias
```
