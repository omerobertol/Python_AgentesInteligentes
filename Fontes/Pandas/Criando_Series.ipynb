```python
import pandas as pd
import numpy as np

numeros = pd.Series([10, 11, 12, 13, 14])
print(type(numeros))
print(numeros)
```

```python
x = pd.Series([i for i in range(1, 11)])
print(x)
```

```python
valores = [4, 5, 4] # idade das crianças
indices = ['Joaquim', 'Ayla', 'Davi'] # crianças

criancas = pd.Series(valores, index=indices)

print(criancas)
```

```python
x = np.array([10, 11, 12, 13, 14])
print(type(x))
print(x)
```

```python
nros = pd.Series(x)
print(type(nros))
print()
print(nros)
print("\nQuantidades de elementos = ", nros.size)
```

```python
valores = [np.random.randint(1, 100) for i in range(10)]
x = pd.Series(valores)

print(type(x))
print()
print(x)
```

```python
valores = np.random.rand(7)
x = pd.Series(valores)

print(type(x))
print()
print(x)
```

```python
dic = {'Omero': 60,
       'Vanessa': 42,
       'Joaquim': 4}

print(type(dic))
print(dic)
print()

pessoas = pd.Series(dic)
print(type(pessoas))
print(pessoas)
```
