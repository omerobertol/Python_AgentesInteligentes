```python
import pandas as pd

# criando um DataFrame a partir de um arquivo 'csv'
df = pd.read_csv('C:\Python_AgentesInteligentes\Fontes\Dados\census.csv')

# Mostra somente os registros com idade menor do que 18
df.loc[(df["age"] < 18) & (df["workclass"] == " Private") & (df['education-num'] == 7) & (df['race'] == " Black")]
```

```python
# Pessoas de 20 a 30 anos e divorciados (campos 'age', 'workclass', 'education-num')
df.loc[(df["age"] >= 20) & (df["age"] <= 30) & 
       (df["marital-status"] == " Divorced"), ['age', 'workclass', 'education-num']]
```

```python
# Pessoas com 30 anos e divorciadas
df.query("age == 18 and `marital-status` == ' Divorced'")
```

```python
# Pessoas com 18 anos e casadas
df.query("age > 18 and sex == ' Female'")
```
